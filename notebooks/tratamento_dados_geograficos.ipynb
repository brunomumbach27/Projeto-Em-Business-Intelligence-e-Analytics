```python
import pandas as pd


arquivo =  r"..\data\daos_geograficos\RELATORIO_DTB_BRASIL_2024_MUNICIPIOS.xls"

# Leitura do arquivo
df = pd.read_excel(arquivo, engine="xlrd", skiprows=6)

display(df)
```

```python

df.columns = [col.strip() for col in df.columns]

# Filtrar apenas RS (UF = 43)
df = df[df['Nome_UF'] == 'Rio Grande do Sul']

# Seleção das colunas relevantes para o estudo
colunas_utilizadas = {
    'UF': 'uf',
    'Nome_UF': 'nome_uf',
    'Região Geográfica Intermediária': 'regiao_intermediaria',
    'Nome Região Geográfica Intermediária': 'nome_regiao_intermediaria',
    'Região Geográfica Imediata': 'regiao_imediata',
    'Nome Região Geográfica Imediata': 'nome_regiao_imediata',
    'Município': 'codigo_municipio',
    'Nome_Município': 'nome_municipio'
}

df = df[list(colunas_utilizadas.keys())]
df = df.rename(columns=colunas_utilizadas)

# Limpeza de dados
df['nome_municipio'] = df['nome_municipio'].str.strip().str.upper()

df['codigo_municipio'] = df['codigo_municipio'].astype(str)

df['uf_sigla'] = 'RS'

# Salvar dataset tratado

df.to_csv(r"..\data\daos_geograficos\dados_tratados\dados_geograficos_RS.csv", index=False)

print("✅ Dados geográficos tratados com sucesso!")
```
